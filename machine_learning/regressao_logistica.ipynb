```python
# Importação de bibliotecas
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix, roc_curve, roc_auc_score
from yellowbrick.classifier import ConfusionMatrix
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
```

```python
# Leitura dos dados da dataframe
dados = pd.read_pickle('../dados/dados_tratados.pkl')
```

```python
# Divisão dos dados entre preditores e alvo
x = dados.drop(columns=['liked'])
y = dados['liked']
```

```python
# Divisão da dataframe entre treinamento e teste
x_treinamento, x_teste, y_treinamento, y_teste = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Instanciação do modelo e seu treinamento
modelo = LogisticRegression(random_state = 42)
modelo.fit(x_treinamento, y_treinamento)
```

```python
# Previsão de teste
previsao = modelo.predict(x_teste)
```

```python
# Precisão, Relatório de Classificação e Matriz de Confusão
print("Precisão:", accuracy_score(y_teste, previsao))
print("Relatório de Classificação:\n", classification_report(y_teste, previsao))
cm = confusion_matrix(y_teste, previsao)
print(cm)
cm = ConfusionMatrix(modelo)
cm.fit(x_treinamento, y_treinamento)
cm.score(x_teste, y_teste)
```

```python
# Gráfico Curva ROC
fpr, tpr, thresholds = roc_curve(y_teste, previsao)
auc = roc_auc_score(y_teste, previsao)
plt.plot(fpr, tpr, label=f'ROC curve (area = {auc:.2f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('Taxa de Falsos Positivos')
plt.ylabel('Taxa de Verdadeiros Positivos')
plt.title('Curva ROC')
plt.legend(loc='lower right')
plt.savefig('../graficos/Gráfico Curva ROC.png', dpi=300)
plt.show()
```

```python
# Exportação do modelo
with open('../dados/meu_modelo.pkl', 'wb') as f:
    pickle.dump(modelo, f)
```
