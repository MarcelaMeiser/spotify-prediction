```python
# Função para identificar outliers e mostrá-los com o gráfico e suas informações
def identificar__outliers(dados):
    colunas_numericas = dados.select_dtypes(include='number').columns
    for coluna in colunas_numericas:
        Q1 = dados[coluna].quantile(0.25)
        Q3 = dados[coluna].quantile(0.75)
        IQR = Q3 - Q1
        limite_inferior = Q1 - 1.5 * IQR
        limite_superior = Q3 + 1.5 * IQR
        outliers = dados[(dados[coluna] < limite_inferior) | (dados[coluna] > limite_superior)]
        print(f"\nColuna: {coluna}")
        print(f"Número de outliers: {outliers.shape[0]}")
        plt.figure(figsize=(6, 4))
        dados.boxplot(column=coluna)
        plt.title(f'Boxplot - {coluna}')
        plt.show()
        if not outliers.empty:
            display(outliers)
        else:
            print("Nenhum outlier encontrado.")
```

```python
# Usar a função de identificar outliers
identificar__outliers(dados)
```

```python
# Dropar Outliers que possam existir a partir da coluna Liked
dados = dados[dados['liked'].isin([0, 1])]
```

```python
# Descrição dos dados da dataframe
dados.describe(include='all')
```

```python
# Gráfico HeatMap para identificar lógicas
dados_numericos = dados.select_dtypes(include='number')
sns.heatmap(dados_numericos.corr(), annot=True)
plt.tight_layout()
plt.savefig('../graficos/Gráfico HeatMap.png', dpi=300)
plt.show()
#Com esse gráfico conseguimos identificar as seguintes lógicas:
#1ª: Quanto maior a popularidade da música, maior a chance da pessoa gostar.
#2ª: Quanto mais longa a música, menor a chance da pessoa gostar.
```
